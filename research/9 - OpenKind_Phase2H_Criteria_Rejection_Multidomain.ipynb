## Goal
Test the open questions in **Whitepaper v0.3 §13** without mixing a new decision contract with an unannounced execution change. The main study compares the old scorer, training-example-backed criteria, small rejection/calibration updates, and multi-domain candidate-head fitting. Strict FP32 remains the reference. Rust/Metal and real concurrent service validation are deferred.

**There are no precomputed Qwen results in this notebook.** Its local validation used tiny CPU models and explicitly synthetic inputs. Real Qwen/CUDA results come from your run.

Use a **fresh L4 or A100 runtime**, enable Colab secret **`HF_TOKEN`**, and run cells in order. Do not load Qwen in the notebook coordinator. Each model worker starts in a separate process. Full FP32 will not fit an ordinary T4 runtime under this study's memory guard.

The default study is larger than Phase 2G because it includes fitting, not just inference regression. `quick` checks a smaller experiment; `smoke` checks plumbing and cannot support quality claims. The optional online/LoRA pair requires at least 32 GiB total GPU memory.

**2h.1.1 repair:** Copy destinations are created at export time; prepared data can be resumed without reserving a new split. Unchanged setup-cell outputs may be retained from the earlier session; the corrected export has not been run on Colab here.


**2h.1.2 maintenance repair:** fixes the undefined memory helper, adds resumable final-output checkpoints, and preserves the original task/loss/selection contracts. Historical outputs have been cleared from the repaired notebook so they cannot be mistaken for new execution. The pre-repair notebook is retained in the repair package.

## Context & methods
| Question | Implemented treatment | Default | Interpretation boundary |
|---|---|---|---|
| Criteria clarity | Original terse names versus reserved author-training examples; optional reviewed definitions | On | Exemplar support is extra labeled input, not a pure paraphrase or independent review |
| Rejection transfer | Frozen scorer + old none; frozen scorer + new none/calibration; jointly fitted multi-domain head + new none | On | Omitted in-scope and author-OOS are separate strata |
| Context/instruction robustness | First/last evidence, unrelated notes, instruction-in-notes control; optional labeled natural documents | On | Constructed contexts are not natural long-document evidence |
| Arithmetic | Same fitted artifacts under strict FP32 and TF32 permission | On | Report numerical, argmax, and policy changes separately |
| Efficient baselines | Character n-gram dynamic scorer and one-step finite-code Qwen logits | On | Same task/splits, but different input rendering and no equal-gradient-budget claim |
| Noul/Score | Supervised BoolQ yes/no and five-level SST sentiment readouts | On | Bounded domain probes, not generic primitives or unknown-evidence handling |
| State-first rendering | Separately fitted heads + related-query prefix mechanics | Optional | New prompt contract; not arbitrary-question semantic proof |
| Smaller backbone | Independently fitted Qwen3.5-2B-Base heads | Optional | Immutable revision resolved before fitting; no assumed winner |
| LoRA | Matched online-head-only versus last-four-block LoRA candidate-pair training | Optional | Same BCE loss/pairs/updates; not directly matched to joint-CE treatment |
| Rust/Metal/service | Head fixtures and manifests exported | Deferred | No GPU-to-Mac performance extrapolation |

### Key assumptions
- All C/D/E/G inspected message groups are excluded, plus prior Phase 2H reservations. This is exact normalized-message exclusion, **not pretraining or paraphrase decontamination**.
- Fitting, development, calibration-fit, calibration-gate, policy-development and final groups are distinct. Criteria examples come from an additional support set.
- The prior **57/20 Banking label partition** is preserved. Six CLINC domains supply fitting labels; **travel and work** are held out from all head fitting and criteria examples.
- Training/selection weights are **60% answer-present, 25% omitted intent, 15% author-OOS**. Policy scenarios use absent priors 5/25/50%, with author-OOS constituting an assumed half of absent cases. These are declared assumptions, not measured traffic.
- Main final tests use K=4/16. Training episodes use K=4/8. Source labels and omissions are never altered after predictions.
- The frozen-control probability model is unchanged, but **every new treatment selects its own application thresholds on the separate policy-development set**. “Criteria-only” refers to model parameters, not frozen historic thresholds. Cross-arithmetic tests do freeze the newly selected policies.
- The final lock is a workflow guard, not protection against manually opening local files. Reviewed final results become historical; do not tune against them and call the next replay fresh.

### 1. Install reference dependencies
This pins the Transformers/data stack without replacing Torch/CUDA. Workers require **PyTorch 2.11.x** and Transformers 5.17.0. A different Torch release is a different environment—not a silent fallback.

````python
# Same non-Torch dependency contract as the source Phase 2H notebook.
import sys, subprocess, importlib.metadata as metadata
from pathlib import Path

if 'torch' in sys.modules and sys.modules['torch'].cuda.is_initialized():
    raise RuntimeError('Use a fresh Colab GPU runtime: the notebook coordinator must not own CUDA state.')
try:
    torch_version = metadata.version('torch')
except metadata.PackageNotFoundError:
    torch_version = '(missing)'
if not torch_version.startswith('2.11.'):
    raise RuntimeError(
        'The recorded study requires PyTorch 2.11.x (source run: 2.11.0+cu128). '
        f'This runtime has {torch_version}. No silent Torch/CUDA replacement is performed. '
        'Use the original compatible Colab runtime/environment before continuing.')

subprocess.check_call([sys.executable, '-m', 'pip', 'install', '--quiet',
    'transformers==5.17.0', 'accelerate==1.14.0', 'datasets==4.8.5',
    'safetensors==0.8.0', 'scikit-learn==1.6.1', 'numpy>=2,<3',
    'scipy>=1.16,<2', 'pandas', 'matplotlib', 'tqdm', 'psutil'])
print('Pinned dependency setup complete. Torch/CUDA were not replaced.')
````

Set `RUN_FINAL=False` for a development-only run. `REQUIRE_REVIEW_FOR_FINAL=True` requires a complete reviewer-supplied definition file; the default support-example study is explicitly **unreviewed**.

Only run one preparation session against the reservation registry at a time. The registry prevents silent reuse of previously reserved H examples; it is not a distributed coordination service. New studies consume new data, and will fail rather than reuse exhausted final pools.

````python
#@title Recovery configuration
SOURCE_ARCHIVE = '/content/drive/MyDrive/Colab Notebooks/OpenKind_Phase2H_results/20260919T040612625670Z/openkind_phase2h_20260919T040612625670Z.zip'
EXPECTED_SOURCE_SHA256 = '51c772ab248d8949fe73bbbd91056826975eebb3594d03ca693438355d9a4cda'
WORK_ROOT = '/content/openkind_phase2h_finish'
DESTINATION_ROOT = '/content/drive/MyDrive/Colab Notebooks/OpenKind_Phase2H_Finish_results'
FINISH_TAG = 'finish_2h_1_2'  # Keep this unchanged to resume this continuation.
SAVE_EVALUATION_FEATURES = True  # SQLite-safe backups in the periodic checkpoint ZIPs.
CHECKPOINT_SECONDS = 120         # Best-effort interval, checked between units of work.

# Source configuration controls data, seeds, criteria, thresholds, precision and arms.
# This notebook does not expose overrides that would silently change the experiment.

NOTEBOOK_MODE = 'finish_saved'  # finish_saved / new_study

from phase2h_config import Settings
PRESET = 'standard'  # standard / quick / smoke
RUN_FINAL = True
RUN_TF32 = True
RUN_PRIMITIVES = True
RUN_FINITE_TOKEN = True
RUN_STATE_FIRST = False
RUN_SMALLER_MODEL = False
RUN_LORA = False  # matched online-head + LoRA controls; >=32 GiB GPU required
REQUIRE_REVIEW_FOR_FINAL = False
CRITERIA_REVIEW_JSON = ''   # complete two-reviewer definition manifest, optional
NATURAL_DOCUMENTS_JSONL = ''  # independently labeled natural documents, optional

cfg = Settings(
    preset=PRESET,
    run_final=RUN_FINAL,
    run_tf32=RUN_TF32,
    run_primitives=RUN_PRIMITIVES,
    run_finite_token=RUN_FINITE_TOKEN,
    run_state_first=RUN_STATE_FIRST,
    run_smaller_model=RUN_SMALLER_MODEL,
    run_lora=RUN_LORA,
    criteria_review_json=CRITERIA_REVIEW_JSON,
    require_independent_review_for_final=REQUIRE_REVIEW_FOR_FINAL,
    natural_documents_jsonl=NATURAL_DOCUMENTS_JSONL,
)
print(json.dumps(cfg.json(), indent=2))
print('Per-family sizes:', cfg.sizes())
````

## Data
### 6. Reserve source-separated messages and version the criteria
The standard decision study reserves **294 criteria-support messages**, then:

| Partition | Banking fitting labels | CLINC fitting domains | Author OOS | Role |
|---|---:|---:|---:|---|
| Train | 256 | 256 | 40 | Candidate / none fitting |
| Development | 64 | 64 | 12 | Checkpoints and none-model choice |
| Calibration fit | 64 | 64 | 12 | Fit temperature |
| Calibration gate | 48 | 48 | 12 | Accept/reject temperature |
| Policy development | 64 | 64 | 12 | Cost/coverage thresholds |

Final: **64 messages each** from Banking seen labels, Banking held-out labels, CLINC fitting domains, CLINC held-out domains, and author OOS: **320 message groups / 1,152 episodes**. Counts of repeated candidate/omission variants are not independent samples.

BoolQ and SST-5 are separately partitioned: each uses 384 train, 64 development, 64 calibration-fit, 64 calibration-gate, and 128 final groups. BoolQ's author validation split supplies its labeled final examples; SST-5 uses the mirror's test split. Over-length inputs are explicitly excluded before selection rather than silently truncated.

The criteria template contains original names, reserved training examples and pending definition/include/exclude fields. The lexical pair queue is built from names, not final errors. Code cannot certify that the named reviewers independently reviewed the definitions.

**Export recovery (2h.1.1):** Preparation checkpoints its existing data before publishing to Drive. If that export fails, restore Drive access and run `data_audit = study.resume_prepare_export()` on the same `study`. Do not construct another experiment or delete the reservation registry. This does not redraw or refit data.


In the default `finish_saved` mode, this section validates and copies the already-reserved inputs. It never calls the data-reservation code or updates the original registry.

````python
from phase2h_workflow import Experiment
from phase2h_finish import FinishExperiment
if NOTEBOOK_MODE == 'finish_saved':
    study = FinishExperiment(SOURCE_ARCHIVE, expected_sha256=EXPECTED_SOURCE_SHA256,
        work_root=WORK_ROOT, destination_root=DESTINATION_ROOT, finish_tag=FINISH_TAG,
        copy_to_drive=True, save_features=SAVE_EVALUATION_FEATURES,
        checkpoint_seconds=CHECKPOINT_SECONDS)
    data_audit = study.prepare()
    cfg = study.cfg  # Source configuration, not edited new-study settings, controls recovery.
elif NOTEBOOK_MODE == 'new_study':
    if 'study' in globals():
        raise RuntimeError('A study object already exists; do not reserve another split by rerunning this cell.')
    study = Experiment(cfg)
    data_audit = study.prepare()
else:
    raise ValueError('NOTEBOOK_MODE must be finish_saved or new_study')
print('Run ID:', study.run_id)
print('Original final rows are not displayed or used for fitting.')
````

## Learning treatments
### 7. Fit without reading final examples
For each criteria variant, the 4B treatment graph is:

1. **Frozen scorer + historical set-none:** no model-parameter fitting (policy thresholds use the new policy-development set).
2. **Frozen scorer + newly fitted none/calibration:** select constant versus set-conditioned none on development data.
3. **Multi-domain candidate-head fitting:** joint candidate-plus-none cross-entropy, seeds 17/29/43, followed by separately fitted none/calibration and policies.

Frozen features are extracted once per finalized token sequence and stored with execution-identity and checksum keys. Only small heads are trained on CPU in the main study. This offline feature cache is bypassed by request timings.

The finite-token baseline scores single answer-code tokens, with none included. It uses no generated explanation. The lexical baseline scores supplied descriptions using fixed-scale character n-gram cosine similarity. Neither receives the same gradient updates as the candidate-head treatment, so report quality–cost trade-offs rather than claiming equal training budgets.

Optional state-first heads use a different registered input order. Optional 2B heads are fitted independently at their own hidden width; enabling this also adds 4B scratch-head controls so the size comparison is not only against a historically trained 4B readout. Optional LoRA is paired with an **online head-only BCE control** using the same pairs, initialization, updates and FP32 storage; both are memory-guarded. Do not use LoRA-versus-joint-CE differences alone as the LoRA effect.

In `finish_saved` mode, `study.fit()` is a compatibility method that validates and reuses the existing artifacts. It does not invoke any fitting, calibration or policy-selection function.

````python
FIT_ERROR = None
try:
    study.fit()
except Exception as exc:
    FIT_ERROR = str(exc)
    study.report['coordinator_fit_error'] = FIT_ERROR
    study.save()
    print('Fitting stopped:', FIT_ERROR)
    print('Partial artifacts are retained. Later export still runs; final requires a completed 4B baseline.')
````

````python
fit_table = []
for name, record in study.report['workers'].items():
    if not name.startswith('fit_'):
        continue
    path = Path(record['report_path'])
    report = json.loads(path.read_text()) if path.exists() else {}
    for profile in report.get('profiles', []):
        fit_table.append({'worker': name, 'profile': profile['name'], 'development weighted NLL': profile['dev_nll']})
if fit_table:
    display(pd.DataFrame(fit_table).sort_values(['worker', 'development weighted NLL']).head(40))
else:
    print('No completed profile table. Inspect the first worker traceback; do not infer model quality from a failed run.')
````

### 8. Seal treatments, then evaluate final data once
Before reading final targets in a worker, the coordinator hashes the final payload, criteria, all fitted heads, temperatures and policy-selection files. No final metric can change those artifacts during this evaluation.

The selected trained profile is chosen by development weighted NLL; every preregistered arm remains in the final report. An apparently better final arm is a finding to confirm, not permission to reuse final data for another selection round. The 0.005 numerical tolerance is unchanged; argmax and directed application actions are checked too.

The context study compares the development-selected head with its matching frozen-criteria control. Added administrative notes and injected instructions are controlled transformations, not naturally annotated long documents. Natural-document status remains `not_provided` unless you supplied independent source labels.

````python
FINAL_ERROR = None
if FIT_ERROR is None:
    try:
        study.evaluate()
    except Exception as exc:
        FINAL_ERROR = str(exc)
        study.report['coordinator_final_error'] = FINAL_ERROR
        study.save()
        print('Final evaluation stopped:', FINAL_ERROR)
        print('This will be reported as partial/development-only, not as an accepted model.')
else:
    print('Final not opened because fitting did not finish.')
````

## Results
### 9. Save reports, fixtures and provenance
Reports retain every treatment, conditional omitted/OOS rates, cost/coverage, message-clustered uncertainty, error dossiers, criteria versions, source hashes and the final lock. No final metric automatically promotes TF32, a new head, or a policy.

The archive includes small head/reference fixtures, not base weights or cached feature SQLite databases. Live feature files remain in the runtime. When enabled, a LoRA adapter is exported separately with exact target names and its checkpoint revision. That is not Rust/Metal parity.

````python
summary = study.finish()
print('Overall execution status:', summary['status'])
print('Unanswered/deferred items:', json.dumps(summary['open_question_status'], indent=2))
if summary['status'] != 'completed':
    print('PARTIAL: the ZIP was exported, but required evaluation work has not completed.')
else:
    print('Execution complete. Inspect quality and numerical acceptance separately.')
````

````python
selected_rows = []
for worker_name, result in summary.get('results', {}).items():
    if not worker_name.startswith('eval_'):
        continue
    for family, metrics in result.get('selected_metrics', {}).items():
        q = metrics['quality']
        selected_rows.append({'worker': worker_name, 'family': family, 'selected before final': result['selected_before_final'],
                              'messages': q['messages'], 'episodes': q['episodes'], 'accuracy': q['accuracy'], 'NLL': q['nll']})
if selected_rows:
    display(pd.DataFrame(selected_rows))
else:
    print('No final quality rows are available. Check status and first traceback.')
````

### Primitive and policy readouts
The primitive table reports the seed selected on development data, not the best final seed. The policy table uses the pooled declared mixture; family-specific costs lacking OOS or answerable cases remain explicitly conditional in the full report.

````python
primitive_rows, policy_rows = [], []
for worker_name, result in summary.get('results', {}).items():
    if not worker_name.startswith('eval_'):
        continue
    for task, entry in result.get('primitive_metrics', {}).items():
        for run in entry['runs']:
            if run['seed'] != entry['selected_seed_before_final']:
                continue
            metric = run['final']
            primitive_rows.append({'worker': worker_name, 'task': task, 'seed': run['seed'],
                'groups': metric['groups'], 'accuracy': metric['accuracy'], 'NLL': metric['nll'],
                'binary Brier': metric.get('binary_brier_yes'), 'expected-level MAE': metric.get('expected_level_mae')})
    pooled = result.get('selected_metrics', {}).get('pooled_declared_mixture', {})
    for policy in pooled.get('policies', []):
        policy_rows.append({'worker': worker_name, 'absent prior': policy['assumed_absent_prior'],
            'wrong cost': policy['wrong_cost'], 'threshold': policy['threshold'],
            'scenario coverage': policy['scenario_coverage'], 'scenario cost': policy['scenario_cost'],
            'accepted episodes': policy['accepted_episodes'], 'wrong accepted': policy['wrong_accepted_episodes']})
if primitive_rows:
    display(pd.DataFrame(primitive_rows))
if policy_rows:
    display(pd.DataFrame(policy_rows))
````
