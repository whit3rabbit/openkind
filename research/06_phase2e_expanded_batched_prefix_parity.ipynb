## 3 · Configuration and source archives

`standard` selects **four seen-label and four held-out-label messages** from the archived E test partitions by a deterministic hash. All sixteen variants per message are included: four candidate counts × two distractor samplers × correct intent present/absent. That is **128 episodes from eight independent messages**.

| Preset | Distinct messages | Parity episodes per mode | Complete-request benchmark episodes | Profile episodes |
|---|---:|---:|---:|---:|
| smoke | 2 | 32 | 16 | 16 |
| quick | 4 | 64 | 16 | 16 |
| standard | 8 | 128 | 32 | 16 |

## 4 · What each isolated worker does

Both workers use identical finalized token sequences and the same frozen parameters/policies.

**Memory and numerical reference.** Each worker starts a new Python interpreter, verifies the installed cache API on a tiny random CPU Qwen, then loads the text model **directly in its target dtype**. FP32 is not obtained by retaining and converting a BF16 GPU model. Before loading, the guard compares actual driver-free memory against the target weight footprint plus a 2 GiB reserve. It records tensor allocations, allocator reservations, and driver-free memory before and after loading and each stage. No CPU offload or silent lower-precision fallback is used.

**Expanded parity.** Every panel episode is evaluated with full sequential prompts, full-prompt batches of four, sequential cached suffixes, and the new suffix-batching prototype. The report compares all three frozen none-head distributions and all nine policies. It separates argmax changes, answer→review, review→answer, and a different accepted candidate. One hard/absent four-candidate episode per message also receives reversed-order and repeat checks.

**Component profiling.** Separate synchronized wall times cover tokenization/planning, input packing/transfer, full forward or prefill, cache cloning/expansion, suffix forward, score transfer/head, and probabilities/policies/JSON. Synchronization is intrusive; these times are **not pure GPU kernel times** and are not substituted for the uninstrumented request benchmark. Unattributed host time is recorded too.

**Prototype.** Suffixes are grouped by exact token length and evaluated in groups of up to four. The entire hybrid cache is cloned and expanded using the supported cache-reordering operation. There is **no suffix padding**, no state aliasing with the reusable root, and scores are restored to original candidate order. This reduces calls only when compatible suffix lengths occur. It remains experimental and receives the same probability/action checks as the baseline.

All cold-request timings include a new prefill and cache-copy work. Diagnostic cache hashes run outside timed calls. Long-prefix timings remain explicitly synthetic, not decision-quality evidence.
